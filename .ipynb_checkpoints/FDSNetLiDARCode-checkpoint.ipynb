```python
from mmdet3d.apis import inference_detector, init_model
from mmdet3d.registry import VISUALIZERS
from mmdet3d.utils import register_all_modules
```

```python
register_all_modules()
```

```python
config_file = '/home/asaad/Documents/2024/FDSNet/mmdetection3d-1.2.0/configs/pointpillars/pointpillars_hv_secfpn_8xb6-160e_kitti-3d-3class.py'
checkpoint_file = '/home/asaad/Documents/2024/FDSNet/mmdetection3d-1.2.0/checkpoints/hv_pointpillars_secfpn_6x8_160e_kitti-3d-3class_20220301_150306-37dc2420.pth'
```

```python
model = init_model(config_file, checkpoint_file, device='cuda:0')
```

```python
# Build visualizer
visualizer = VISUALIZERS.build(model.cfg.visualizer)
visualizer.dataset_meta = {
    'CLASSES': model.cfg.class_names,
    # Optionally define a palette if needed:
    'PALETTE': [
        (255, 158, 0),   # Car
        (0, 255, 0),     # Pedestrian
        (0, 0, 255)      # Cyclist
    ]
}
```

```python
# test a single sample
pcd = '/home/asaad/Documents/2024/FDSNet/mmdetection3d-1.2.0/data/nuscenes/n015-2018-07-24-11-22-45+0800__LIDAR_TOP__1532402927647951.pcd.bin'
result, data = inference_detector(model, pcd)
points = data['inputs']['points']
data_input = dict(points=points)
```

```python
# show the results
out_dir = ''
visualizer.add_datasample(
    'result',
    data_input,
    data_sample=result,
    draw_gt=False,
    show=True,
    wait_time=10,
    out_file=out_dir,
    vis_task='lidar_det')
```
